# Running sample: wprp based on mass

```python
i = 0
galaxies = f"/projects/hywu/cluster_sims/cluster_finding/data/emulator_data/base_c000_ph{i:03d}/z0p300/model_hod000000/gals.fit"
halos = f"/projects/hywu/cluster_sims/cluster_finding/data/AbacusSummit_base/base_c000/base_c000_ph{i:03d}/z0p300/halos_3e+12.fit"
cluster_data = f"/projects/hywu/cluster_sims/cluster_finding/data/emulator_data/base_c000_ph{i:03d}/z0p300/model_hod000000/richness_q180_bg.fit"
current = WpCalculations()
print("cluster")
wpc, rpc, massc = current.error_wp_calc_oldv(galaxies, cluster_data, clusters = True, mass = True)
print("halos")
wph, rph, massh = current.error_wp_calc_oldv(galaxies, halos, clusters = False, mass = True)
```

```python
plt.plot(rpc, wpc, label = "clusters")
plt.plot(rph, wph, label = "halos")
plt.legend()
plt.title("*cluster wprp on mass counts")
plt.loglog()
```

```python
plt.hist(np.log10(massc), bins = 200, alpha = 0.5, label = "clusters")
plt.hist(np.log10(massh), bins = 200, alpha = 0.5, label = "halos")
plt.legend()

plt.yscale('log')
```

# Running sample: comparing cluster richness vs. mass data

```python
current = WpCalculations()
wpcr, rpcr, masscr = current.error_wp_calc_oldv(galaxies, cluster_data, clusters = True, mass = False)
```

```python
plt.plot(rpc, wpc, label = "clusters mass")
plt.plot(rph, wph, label = "halos mass")
plt.plot(rpcr, wpcr, label = "clusters richness")
plt.legend()
plt.loglog()
```

```python
plt.hist(np.log10(massc), bins = 200, alpha = 0.5, label = "clusters mass")
plt.hist(np.log10(masscr), bins = 200, alpha = 0.5, label = "clusters richness")
plt.hist(np.log10(massh), bins = 200, alpha = 0.5, label = "halos mass")
print(len(massc), len(masscr), len(massh))
plt.legend()
plt.yscale('log')
```
