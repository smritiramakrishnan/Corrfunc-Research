# Cross correlational wp calculations: Github function

## Cross correlation calculation with 20 bins

### M200m > 10^14

```python
nbins = 20
rmin = 0.1
rmax = 20.0
# rpavg for M200m > 10^14
rbins = np.logspace(np.log10(0.1), np.log10(rmax), nbins + 1)
cross_corr_wp, cross_corr_rpavg = ct.wp_pairs_cross(g_x, g_y, g_z, h_x14, h_y14, h_z14, pimax = 100, bins = rbins, boxsize = max(g_x))
pg.plot_results([cross_corr_rpavg], [cross_corr_wp])
```

```python
pg.plot_results([galaxy_results[1], cross_corr_rpavg], [galaxy_results[0], cross_corr_wp], ['galaxy wp', 'cross corr wp']) 
```

### M200m > 10^13

```python
# rpavg for M200m > 10^13
cross_corr_wp, cross_corr_rpavg = ct.wp_pairs_cross(g_x, g_y, g_z, h_x13, h_y13, h_z13, pimax = 100, bins = rbins, boxsize = max(g_x))
pg.plot_results([cross_corr_rpavg], [cross_corr_wp])
```

```python
pg.plot_results([galaxy_results[1], cross_corr_rpavg], [galaxy_results[0], cross_corr_wp], ['galaxy wp', 'cross corr wp'])
```
