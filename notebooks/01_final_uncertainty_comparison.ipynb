```python
import numpy as np
from Corrfunc.theory import wp
from Corrfunc.theory.DDrppi import DDrppi
import matplotlib.pyplot as plt

from dotenv import load_dotenv
import sys
import os
sys.path.append(os.path.abspath(os.path.join('..')))
import src.correlation_tools as ct
import src.plotting as pg

load_dotenv()
data_dir = os.getenv('data_dir')
```

# Mass > 10**14

```python
meanrp_14_100p_ss = []
frac_err_14_100p_ss = []

colors = ['blue', 'orange', 'green', 'red']
data_14_100p_0s = np.load(f'{data_dir}mass_e14.npz')
rpavg_14_100p_0s = data_14_100p_0s['array_one']
wp_14_100p_0s = data_14_100p_0s['array_two']
mean_rp_14_100p_0s, mean_wp_14_100p_0s, stdev_14_100p_0s = ct.error_data(rpavg_14_100p_0s, wp_14_100p_0s)
```

```python
meanrp_14_100p_ss.append(mean_rp_14_100p_0s)
frac_err_14_100p_ss.append(pg.get_frac_err(stdev_14_100p_0s, mean_wp_14_100p_0s))

plt.plot(mean_rp_14_100p_0s, mean_wp_14_100p_0s, label = 'scale = 0')
scales = [30, 60, 100]
for scale in scales:
    data = np.load(f'{data_dir}uncertainty_14_{scale}.npz')
    rp_ = data['array_one']
    wp_ = data['array_two']
    mean_rp, mean_wp, stdev = ct.error_data(rp_, wp_)
    frac_err_14_100p_ss.append(pg.get_frac_err(stdev, mean_wp))
    meanrp_14_100p_ss.append(mean_rp)
    plt.plot(mean_rp, mean_wp, label = 'scale = ' + str(scale))

plt.loglog()
plt.xlabel(r'$\langle r_p\rangle$')
plt.ylabel(r'$w_p$')
plt.title(r'$w_p$ vs $\langle r_p\rangle$')
plt.legend(fontsize = 'small')
#plot_frac_err(meanrp_14_100p_ss, frac_err_14_100p_ss, ['scale = 0', 'scale = 30', 'scale = 60', 'scale = 100'], colors, y_max = 0.1)
#plt.title('Fractional error vs Mean rp_avg: Mass > 10**14, variable scale')
```

```python
pg.plot_frac_err(meanrp_14_100p_ss, frac_err_14_100p_ss, ['scale = 0', 'scale = 30', 'scale = 60', 'scale = 100'], colors)
```

```python
meanrp_14_pp_0s = []
frac_err_14_pp_0s = []

meanrp_14_pp_0s.append(mean_rp_14_100p_0s)
frac_err_14_pp_0s.append(pg.get_frac_err(stdev_14_100p_0s, mean_wp_14_100p_0s))

percents = ['10', '1', '001']

plt.plot(mean_rp_14_100p_0s, mean_wp_14_100p_0s, label = '100%')

for p in percents:
    data = np.load(f'{data_dir}subsamp_14_{p}_0.npz')
    rp_ = data['rp']
    wp_ = data['wp']
    mean_rp, mean_wp, stdev = ct.error_data(rp_, wp_)
    if p == '001':
        plt.plot(mean_rp, mean_wp, label = '0.1%')
    else:
        plt.plot(mean_rp, mean_wp, label = p + '%')
    frac_err_14_pp_0s.append(pg.get_frac_err(stdev, mean_wp))
    meanrp_14_pp_0s.append(mean_rp)

plt.loglog()
plt.xlabel(r'$\langle r_p\rangle$')
plt.ylabel(r'$w_p$')
plt.title(r'$w_p$ vs $\langle r_p\rangle$')
plt.legend(fontsize = 'small')
```

```python
pg.plot_frac_err(meanrp_14_pp_0s, frac_err_14_pp_0s, ['100%', '10%', '1%', '0.1%'], colors, y_max = 0.1)
```

```python
data_14_100p_30s = np.load(f'{data_dir}uncertainty_14_30.npz')
rpavg_14_100p_30s = data_14_100p_30s['array_one']
wp_14_100p_30s = data_14_100p_30s['array_two']
mean_rp_14_100p_30s, mean_wp_14_100p_30s, stdev_14_100p_30s = ct.error_data(rpavg_14_100p_30s, wp_14_100p_30s)

meanrp_14_pp_30s = []
frac_err_14_pp_30s = []

meanrp_14_pp_30s.append(mean_rp_14_100p_30s)
frac_err_14_pp_30s.append(pg.get_frac_err(stdev_14_100p_30s, mean_wp_14_100p_30s))

percents = ['10', '1', '001']
for p in percents:
    data = np.load(f'{data_dir}subsamp_14_{p}_30.npz')
    rp_ = data['rp']
    wp_ = data['wp']
    mean_rp, mean_wp, stdev = ct.error_data(rp_, wp_)
    frac_err_14_pp_30s.append(pg.get_frac_err(stdev, mean_wp))
    meanrp_14_pp_30s.append(mean_rp)

pg.plot_frac_err(meanrp_14_pp_0s, frac_err_14_pp_0s, [], colors)
pg.plot_frac_err(meanrp_14_pp_30s, frac_err_14_pp_30s, ['100%', '10%', '1%', '0.1%'], colors, dotted=True, y_max = 0.1)
plt.title('Fractional error vs Mean rp_avg: Mass > 10**13 (dotted line = scale of 30)')

```

```python
meanrp_13_100p_ss = []
frac_err_13_100p_ss = []

data_13_100p_0s = np.load(f'{data_dir}mass_e13.npz')
rpavg_13_100p_0s = data_13_100p_0s['array_one']
wp_13_100p_0s = data_13_100p_0s['array_two']
mean_rp_13_100p_0s, mean_wp_13_100p_0s, stdev_13_100p_0s = ct.error_data(rpavg_13_100p_0s, wp_13_100p_0s)

meanrp_13_100p_ss.append(mean_rp_13_100p_0s)
frac_err_13_100p_ss.append(pg.get_frac_err(stdev_13_100p_0s, mean_wp_13_100p_0s))

scales = [30, 60, 100]
for scale in scales:
    data = np.load(f'{data_dir}uncertainty_13_{scale}.npz')
    rp_ = data['array_one']
    wp_ = data['array_two']
    mean_rp, mean_wp, stdev = ct.error_data(rp_, wp_)
    frac_err_13_100p_ss.append(pg.get_frac_err(stdev, mean_wp))
    meanrp_13_100p_ss.append(mean_rp)


pg.plot_frac_err(meanrp_13_100p_ss, frac_err_13_100p_ss, ['scale = 0', 'scale = 30', 'scale = 60', 'scale = 100'], colors, y_max = 0.045)
plt.title('Fractional error vs Mean rp_avg: Mass > 10**13, variable scale')
```
