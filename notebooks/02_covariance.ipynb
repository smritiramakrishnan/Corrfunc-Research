```python
import numpy as np
import matplotlib.colors as mplcolor
import seaborn as sns

import os
from dotenv import load_dotenv
import sys

sys.path.append(os.path.abspath(os.path.join('..')))
import src.correlation_tools as ct

load_dotenv()
data_dir = os.getenv('data_dir')
```

```python
# Going through 
file = f"{data_dir}wp_ph000.npz"
data = np.load(file)
all_wp = data['wp_']
all_rp = data['rp_avg']
wp_means = []
__r, wp_m, __s = ct.error_data(all_rp, all_wp)
wp_means.append(wp_m)
for i in range (1, 25):
    file = f"{data_dir}wp_ph{i:03d}.npz"
    data = np.load(file)
    all_wp = np.concatenate((all_wp, data['wp_']))
    all_rp = np.concatenate((all_rp, data['rp_avg']))
    rp_temp, wp_temp, stdev_temp = ct.error_data(data['rp_avg'], data['wp_'])
    wp_means.append(wp_temp)
print(len(wp_means))
print(type(all_wp))
```

```python
cov_matrix = np.cov(all_wp, rowvar = False)
sns.heatmap(cov_matrix, norm = mplcolor.LogNorm())
```

```python
cov_matrix_25 = np.cov(wp_means, rowvar = False)
sns.heatmap(cov_matrix_25, norm = mplcolor.LogNorm())
```
