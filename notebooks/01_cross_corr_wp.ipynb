```python
import numpy as np

from dotenv import load_dotenv
import sys
import os
sys.path.append(os.path.abspath(os.path.join('..')))
from src.correlation_tools import wp_pairs_cross, create_data
from src.plotting import plot_results
```

```python
load_dotenv()

galaxies = os.getenv('galaxy_test')
halos = os.getenv('halo_test')
```

```python
nbins = 20
rmin = 0.1
rmax = 20.0
rbins = np.logspace(np.log10(rmin), np.log10(rmax), nbins + 1)
```

```python
g_x, g_y, g_z = create_data(galaxies)
h_x, h_y, h_z = create_data(halos)
h_x13, h_y13, h_z13 = create_data(halos, var = 'M200m', min_ = 10**13)
h_x14, h_y14, h_z14 = create_data(halos, var = 'M200m', min_ = 10**14)
```

```python
wp_, rpavg_ = wp_pairs_cross(g_x, g_y, g_z, h_x, h_y, h_z, pimax = 100, bins = rbins, boxsize=max(g_x))
plot_results([rpavg_], [wp_])
```
